```python
def es_primo_manera2(n):
    if n < 2:
        return False

    es_primo = True
    d = 2

    while d * d <= n and es_primo:
        if n % d == 0:
            es_primo = False
        else:
            d += 1

    return es_primo

print("¿91 es primo?:", es_primo_manera2(91))
print("¿97 es primo?:", es_primo_manera2(97))
```
